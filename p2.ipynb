```python
gift_price = 5000

my_money = int(input("Enter the amount: "))

friend_money = int(input("Enter the amount: "))

total_amount = my_money + friend_money

if total_amount >= gift_price and (my_money > 0 or friend_money > 0):
    print("Can purchase the gift I want")
elif total_amount < gift_price and total_amount > 0:
    print("Can purchase some other gift")
else:
    print("Cannot purchase the gift I want")
```

```python
correct_password = "1234"
max_attempts = 3

for attempt in range(max_attempts):
    password = input("Enter your ATM password: ")

    if password == correct_password:
        print("Password correct. Transaction completed.")
        break
    else:
        remaining_attempts = max_attempts - attempt - 1
        print("Wrong password.")

        if remaining_attempts > 0:
            print("Attempts remaining:", remaining_attempts)
        else:
            print("Maximum attempts reached.")
            print("Your card and account have been blocked.")
else:
    print("Please contact your bank.")
```
